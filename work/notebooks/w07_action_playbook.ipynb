## 2. Intended use and limits

### Intended Use Cases
- **Prioritization Aid for Editorial Teams**: This ranked queue is designed to act as a **decision-support tool** to help content managers prioritize which pages to manually inspect and refresh first. It reduces the search space of declining content.
- **Early Risk Identification**: It flags pages experiencing performance decline before the loss becomes catastrophic.

### Strategic Limits and Out-of-Scope Activities
- **No Automatic Publishing**: This queue is not a self-executing automation tool. It does not replace human editorial, styling, or brand voice judgment.
- **No Search Algorithm Claims**: We do not claim this model 'predicts' or 'decodes' search engine algorithms. The model represents observed outcomes inside this specific client portfolio during this time period.
- **No Causal Guarantees**: Refreshes prioritized by this model are not guaranteed to cause performance increases. Causal impact can only be honestly verified via matched control group designs.

## 3. Human review + the no-go list

### Recommended Pre-Refresh Checklist
Before any editor updates a prioritized page, they must verify:
1. **User Intent Fit**: Does the recommended intent (e.g., transactional, informational) match what a human searcher expects?
2. **Information Freshness**: Is the existing page outdated, or is the performance drop caused by external market shifts (such as a competitor launching a superior tool)?
3. **Seasonal Demand**: Is the decline a temporary seasonal trend rather than a permanent loss of visibility?

### The No-Go List
Never refresh or automate changes on pages that fall into these buckets:
- **Recent Refreshes**: Do not update if the page was modified within the last 30 days; allow time for natural search index updates.
- **Stable High-Performers with Declining Demand**: If a page retains #1 rank but traffic drops due to declining industry search volume, a refresh will not recover traffic and might damage its current stable rank.

```python
# Check for fresh pages (updated in < 30 days) that should be excluded from automatic queue actions
if 'df_queue' in globals():
    fresh_pages = df_queue[df_queue['days_since_last_update'] < 30]
    print(f"Number of pages modified in < 30 days (potential No-Go): {len(fresh_pages)}")
```

## 4. Monitoring / retrain triggers

### Performance Monitoring
To ensure our model remains valid and stable, we track the following signals over time:
- **Precision at K (P@K)**: The proportion of flagged pages that actually demonstrate real performance decay.
- **Base Rate of Portfolio Decline**: Monitoring whether the overall rate of natural content decay changes significantly over time.

### Retraining Triggers
We recommend retraining the classifier when we observe:
- **Model Drift**: If the distribution of predictive scores shifts significantly from the baseline training distribution.
- **Performance Degradation**: If precision or ROC-AUC on newly labeled monthly out-of-sample data drops by more than 5% absolute points.
- **Major Search Layout Changes**: If major shifts in search engine results layouts occur, creating structural changes in baseline click-through rates.
