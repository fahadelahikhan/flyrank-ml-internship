## 1. My lane as an ML task (type)

### Task Type Choice: Scoring and Ranking (Priority Score)

* **Why Scoring and Ranking?**
  * The core business problem of content management isn't just knowing if a page is declining in isolation, but deciding **which pages should be prioritized first** for content refresh from a massive list of thousands of pages.
  * Editorial resources are highly constrained. We cannot refresh all declining pages at once. Therefore, we map this problem onto a **Scoring/Ranking task** where each page is assigned a continuous priority score.
  * This allows us to sort the queue of pages descending by priority score so that an editor can work from the top down to maximize recovery value.

```python
import pandas as pd
# Load dataset and calculate baseline proportions to back the ranking/scoring task argument
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
total_pages = len(df)
declining_pages = (df['trend_direction'] == 'down').sum()
declining_pct = (declining_pages / total_pages) * 100

print(f"Total Pages in Portfolio: {total_pages:,}")
print(f"Declining Pages (potential candidates): {declining_pages:,} ({declining_pct:.2f}%)")
print("Clearly, an editor cannot refresh thousands of pages at once; scoring/ranking is essential to prioritize the list.")
```

## 2. Target or proxy

### Target Definition and Source:
* **The Target:** We define our target label `is_declining_label` as a binary flag indicating whether the page's search trend direction is downward (`trend_direction == 'down'`).
* **Is it observed or defined?** It is an **observed outcome** derived from real traffic differences over time (`impressions_last_30d` versus `impressions_prev_30d`), rather than a human-defined subjective rule or manual annotation.
* **Feature Leakage Prevention:** Because our target is derived directly from the trend columns, `trend_direction` and `trend_pct` are strictly excluded from our feature set to ensure an honest, leakage-free modeling framework.

```python
# Verify target distribution and demonstrate no missing labels
target_counts = df['trend_direction'].value_counts(dropna=False)
print("Target Column 'trend_direction' distribution:")
print(target_counts)

# Construct the binary target and verify
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)
print(f"\nBinary target class balance:\n{df['is_declining_label'].value_counts(normalize=True) * 100}")
```

## 3. Success metric

### Metrics and Defensibility:
* **Primary Model Metric:** **ROC-AUC (Receiver Operating Characteristic - Area Under Curve)** and **Precision@K (Precision at top K prioritizations)**.
  * **ROC-AUC** is robust against class imbalances and measures the model's overall capability to rank a randomly chosen declining page higher than a stable/upward page.
  * **Precision@K** directly translates to business value: out of the top `K` pages recommended to editors, what percentage are truly declining? For instance, with `K = 100`, we aim for a Precision@100 significantly exceeding the baseline random rate (~38.4%).

```python
# Establish the random guessing/baseline rate for Precision@K
base_rate = df['is_declining_label'].mean()
print(f"Baseline Precision (random guessing rate): {base_rate * 100:.2f}%")
print("Our success threshold is to achieve a Precision@100 that is at least 1.5x of this baseline rate.")
```

## 5. Why ML beats a fixed rule here

### Why ML over Hardcoded If-Statements?
1. **High Dimensional Interaction:** A page's decline isn't just about high `avg_position` or low `word_count` individually; it is a complex, non-linear interaction of multiple signals (freshness, search volume, competitive level, cpc, scroll rates, and past performance differences).
2. **Client Heterogeneity:** Different clients have completely different traffic scales and page characteristics. Writing manual rules that generalize across 32 clients is virtually impossible without overfitting or underperforming.
3. **Adaptability over Time:** SEO and search intent dynamics shift constantly. An ML model can automatically adapt feature weights as new trailing-window snapshot data arrives.

```python
# Show complexity by highlighting client variance in average word counts and positions
client_stats = df.groupby('client_id').agg(
    mean_word_count=('word_count', 'mean'),
    mean_position=('avg_position', 'mean'),
    decline_rate=('is_declining_label', 'mean')
).head(10)
print("Client-level variation demonstrating why global fixed rules fail:")
print(client_stats)
```
