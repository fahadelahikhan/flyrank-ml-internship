## 1. My lane (or freestyle) and why

**Lane Chosen:** Content Refresh Optimization (Predicting Traffic Decay vs. Freshness)

**Why this lane?** Keeping content fresh is one of the highest ROI marketing decisions. Rather than constantly producing new keyword articles, companies can reclaim decaying search ranking and traffic by updating outdated material. By using machine learning to predict which pieces of content are in active decay based on freshness tiers, age, and historical search impressions, we can systematically recommend exact articles for writers to refresh before major traffic loss occurs.

```python
# Show the count of articles by their current freshness tier and their trend directions to justify the lane
freshness_summary = df.groupby(['freshness_tier', 'trend_direction']).size().unstack(fill_value=0)
display(freshness_summary)
```

## 2. The question: decision, action, cost of a wrong call

* **What decision does this improve?** It improves the decision of *which* content assets to allocate optimization/rewriting budget to.
* **Who acts on it?** SEO managers, content strategists, and freelance copywriters.
* **What does a wrong recommendation cost?**
  * *False Positive (recommending a refresh that isn't needed):* Wastes copywriter fees and content resources on stable pages.
  * *False Negative (failing to recommend a refresh on decaying content):* Costs thousands of dollars in lost organic traffic, lead generation, and eventual revenue as organic impressions plummet.

```python
# Calculate current traffic impact of decaying pages (trend_direction == 'down')
decaying_df = df[df['trend_direction'] == 'down']
total_lost_clicks_potential = decaying_df['clicks_90d'].sum()
print(f"Total 90-day clicks currently sitting in 'down' trend columns: {total_lost_clicks_potential:,} clicks.")
print(f"This represents {len(decaying_df) / len(df) * 100:.1f}% of our total article portfolio which is currently decaying.")
```

## 3. Quick look at the data (2-3 real numbers)

Based on the starter CSV execution, we found three critical data points that prove this lane is highly valuable:
1. **High Decay Prevalence:** 16,262 articles (54.21% of the portfolio) are currently in a downward traffic trend (`trend_direction == 'down'`).
2. **Aging Content Risk:** The average content age is approximately 256 days, indicating that many articles may require review or refreshing to maintain performance.
3. **Freshness Opportunity:** More than half of the articles are currently in a downward traffic trend, and the average content age exceeds 250 days, suggesting that content freshness may be an important factor worth investigating.

```python
# Real numbers from the dataset to back up our 7-week lane:
num_decaying = len(df[df['trend_direction'] == 'down'])
pct_decaying = (num_decaying / len(df)) * 100
mean_age = df['content_age_days'].mean()

print(f"1. Number of decaying articles: {num_decaying:,} ({pct_decaying:.2f}% of portfolio)")
print(f"2. Average age of articles: {mean_age:.1f} days")
print(f"3. Total 90-day impressions at risk: {df[df['trend_direction'] == 'down']['impressions_90d'].sum():,}")
```
