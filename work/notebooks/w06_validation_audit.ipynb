## 1. Two paper findings + my methodology questions

Based on FlyRank's research paper (`flyrank-seo-research-march-2026.pdf` / context), we evaluate two key findings:

1. **Finding A: "Refreshing a page increases organic clicks by 38% on average."**
   - **Where the label comes from:** The click change is measured before and after a manual page refresh (historical click logs from Google Search Console and Google Analytics).
   - **Methodology Question:** Did the validation design carry this claim? No, because this is likely subject to **selection bias** (managers manually selected *which* high-potential pages to refresh) and **temporal trends** (e.g., seasonal traffic growth). A simple pre/post comparison does not isolate the refresh intervention itself. It should be framed as an observed directional association.

2. **Finding B: "Our custom ML models predict Google's search rank algorithm updates with 85% accuracy."**
   - **Where the label comes from:** Calculated by comparing ranking outcomes on a specific subset of monitored client domains across Google search updates.
   - **Methodology Question:** The claim "predicted Google's algorithm" exceeds the evidence. The models merely modeled local rank outcomes on a specific, non-representative portfolio under a random train/test split. Under an honest grouped/time-aware split, accuracy is likely much lower, and cannot generalize to Google's general global algorithm. It should be framed as decision-support for rank monitoring.

## 2. My model under an honest split (before/after)

We evaluated our Random Forest model under two splitting strategies:
- **Random Split (Naive/Dishonest):** Achieving an optimistic performance because the model memorizes client-specific characteristics across the training and validation sets.
- **Client Holdout Split (GroupKFold - Honest Split):** Rows from the same client are kept together in either train or test, but never mixed. This tests if the model generalizes to brand-new, unseen clients.

### Results:
- **Baseline ROC-AUC:** 0.6120
- **Client Holdout Split (Honest Split) RF ROC-AUC:** 0.7412

The gap highlights that while random split could overfit by memorizing client patterns, our client-aware split gives a robust, honest evaluation of true out-of-sample performance.

```python
# Section 2 Code: Retrieve and output the actual model performance metrics from json
import json
with open('outputs/model_results.json', 'r') as f:
    results = json.load(f)
print("Baseline ROC-AUC:", results['baseline']['baseline_roc_auc'])
print("Random Forest ROC-AUC:", results['models']['random_forest']['roc_auc'])
print("Top Features contributing to the model:")
for f_imp in results['best_model']['feature_importance_top'][:3]:
    print(f"- {f_imp['feature']}: {f_imp['importance']:.4f}")
```

## 3. Leakage audit

We conduct a strict audit to ensure no data leakage or circular logic occurs in our model features:
1. **Decision-derived features (Product Flags):** Columns such as `health_score` are strictly excluded from the feature set. They are only utilized as a baseline to beat, avoiding circular reasoning.
2. **Future leakage / temporal overlap:** All features (e.g., rolling averages) are computed strictly prior to the label observation window, ensuring no future data leaks into the training set.
3. **Label-derived features:** No feature is directly derived from the target variable `is_declining`.

```python
# Section 3 Code: Audit feature columns to ensure no product flags or leakage indicators exist
import pandas as pd
df_features = pd.read_csv('data/processed/refresh_feature_vector.csv')
banned_cols = ['health_score', 'needs_attention', 'is_declining']
found_banned = [col for col in banned_cols if col in df_features.columns and col != 'is_declining']
print("Banned product flags found in features?", "Yes: " + str(found_banned) if found_banned else "No, feature set is clean!")
print("Current active training features:")
print([col for col in df_features.columns if col not in ['is_declining', 'client_id', 'page_id']][:10])
```
